```python
import pandas as pd
import numpy as np
```

```python
test_dataframe = pd.read_csv("../test_dataframe.csv")
predictions_check = pd.read_csv("../predictions/predictions.csv")
```

```python
predictions_final = np.load("../predictions/sgd_predictions.npy")
predictions_persistence_final = np.load("../predictions/persistence_predictions.npy") 
```

```python
error_analysis = pd.DataFrame({"timestamp": test_dataframe["timestamp"].values, "actual_cpu": test_dataframe["aggregate_cpu_utilization"].values, "sgd_prediction": predictions_final, "persistence_prediction": predictions_persistence_final})

error_analysis["sgd_error"] = (error_analysis["sgd_prediction"] - error_analysis["actual_cpu"])

error_analysis["persistence_error"] = (error_analysis["persistence_prediction"] - error_analysis["actual_cpu"])

error_analysis["sgd_absolute_error"] = error_analysis["sgd_error"].abs()

error_analysis["persistence_absolute_error"] = (error_analysis["persistence_error"].abs())

print(error_analysis)
```

```python
largest_5 = error_analysis.nlargest(5,"actual_cpu")[["timestamp","actual_cpu","sgd_prediction","persistence_prediction","sgd_absolute_error","persistence_absolute_error"]]
print(largest_5)
```
